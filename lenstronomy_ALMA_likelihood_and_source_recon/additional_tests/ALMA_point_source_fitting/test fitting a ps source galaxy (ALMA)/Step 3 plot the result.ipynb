```python
# the index means [sim number, sample, ra_image and dec_image params], there are 10 simulations and 10 params, might be confusing
chain_list = np.zeros((10, 30000, 10))

for sim in range(10):
    with h5py.File('fitting results//chain_fitting_ps_source_sim' + str(sim) + '.h5', 'r') as f:
        chain_mcmc = f['chain'][:]
    chain_list[sim] = chain_mcmc
```

```python
def correct_prediction_for_100intervals(chain_single_param, param_correct):
    quantile_x = np.linspace(0,1,101)
    prediction = np.zeros((101))

    for i in range(101):
        half_bar = 0.5 * quantile_x[i]
        param_interval_temp = np.zeros((2))
        param_interval_temp = np.quantile(chain_single_param, (0.5 - half_bar, 0.5 + half_bar))
        count_on_errorbar_temp = 0
        if (param_correct > param_interval_temp[0]) and (param_correct < param_interval_temp[1]):
            count_on_errorbar_temp = 1
        prediction[i] = count_on_errorbar_temp
    return prediction
```

```python
quantile_lens_param = np.zeros((101))
n_count_lens_param = 0
for i in tqdm(range(10)):
    for j in range(8):
    
        quantile_lens_param += correct_prediction_for_100intervals(
            chain_list[i,:,j], correct_param_lens[j])
        n_count_lens_param += 1
    
quantile_lens_param /= n_count_lens_param
```

```python
quantile_ps_ra_dec_source = np.zeros((101))
n_count_ps_ra_dec_source = 0
for i in tqdm(range(10)):
    for j in range(2):
    
        quantile_ps_ra_dec_source += correct_prediction_for_100intervals(
            chain_list[i,:,8+j], correct_param_ra_dec_source[j])
        n_count_ps_ra_dec_source += 1
    
quantile_ps_ra_dec_source /= n_count_ps_ra_dec_source
```

Now we use the chain of the independent linear parameters and solved nonlinear parameters to solve for the amps of the point source
- Note that the 1sigma error bars of amplitudes are **not** estimated by linear solving the amplitude of each sample and taking the quantile.
- Instead, we treat each sample point as an independent gaussian likelihood function of the amplitudes, and compute the average and uncertainty using
$$\mu = \frac{1}{N_{sample}}\sum \mu_i$$
$$\sigma^2 = \frac{1}{N_{sample}}\sum(\sigma^2_i+\mu_i^2) - \mu^2$$
where $i$ stands for the sample index.
- $\mu_i$ is the average amplitude for each sample, and is the `params_coefficient` output of the linear solver.
- $\sigma_i^2$ is the variance of the amplitude for each sample, and is the diagonal entries of the `param_cov` output of the linear solver.

```python
from scipy.stats import norm
def sigma_from_percentile(p):
    return norm.ppf((1 + p) / 2)
```

```python
def kwargs_result_embed(sample):
    kwargs_lens = [{ "theta_E": sample[0], "gamma": sample[1], "e1": sample[2], "e2": sample[3], "center_x": sample[4], "center_y": sample[5]},
                   {"gamma1": sample[6],  "gamma2": sample[7], 'ra_0': 0, 'dec_0': 0,} ]
    kwargs_ps = [{"ra_source": sample[8], "dec_source": sample[9], "amp_source": 1}]
    return kwargs_lens, kwargs_ps
```

```python
from lenstronomy.ImSim.image_linear_solve import ImageLinearFit
lens_model_list = ["EPL", "SHEAR"]
lens_model_class = LensModel(lens_model_list=lens_model_list)
point_source_model_list = ['SOURCE_POSITION']
point_source_class = PointSource(point_source_type_list=point_source_model_list, fixed_magnification_list=[True])
```

```python
N_sample = 1000

amp_sigma = np.zeros((10))
amp_average = np.zeros((10))

for sim in tqdm(range(10)):
    data_class.update_data(data_image_set[sim])
    imageLinearFit = ImageLinearFit(data_class, psf_class, lens_model_class=lens_model_class, point_source_class=point_source_class)
    for i in range(N_sample):
        kwargs_lens, kwargs_ps = kwargs_result_embed(chain_list[sim, 20000+i])
        _, _, param_cov, params_coefficient = imageLinearFit.image_linear_solve(
            kwargs_lens = kwargs_lens, kwargs_ps = kwargs_ps, inv_bool=True)

        amp_average[sim] += params_coefficient[0]
        amp_sigma[sim] += param_cov[0,0] + params_coefficient[0]**2

    amp_average[sim] /= N_sample
    amp_sigma[sim] = amp_sigma[sim] / N_sample - amp_average[sim]**2
    amp_sigma[sim] = np.sqrt(amp_sigma[sim])
```

```python
quantile_x = np.linspace(0,1,101)

quantile_amp_ps_source = np.zeros((101))
for j in range(101):
    k = sigma_from_percentile(quantile_x[j])
    prediction = 0
    for sim in range(10):
        if (amp_average[sim] - k * amp_sigma[sim] < 0.002) and (
            amp_average[sim] + k * amp_sigma[sim] > 0.002):
            prediction += 1
    quantile_amp_ps_source[j] = prediction / 10
```

```python
quantile_x = np.linspace(0,1,101)

fig = plt.figure(figsize=(6.3,6))
plt.plot(quantile_x, quantile_x,'--',color='gray',label='Correct statistics\nreference line')
plt.plot(quantile_x, quantile_lens_param, label='Lens params')
plt.plot(quantile_x, quantile_ps_ra_dec_source, label='PS source position')
plt.plot(quantile_x, quantile_amp_ps_source, label='PS source amplitude')
plt.xlim([0, 1.01])
plt.ylim([0, 1.01])
plt.gca().set_yticklabels([f'{x:.0%}' for x in plt.gca().get_yticks()]) 
plt.gca().set_xticklabels([f'{x:.0%}' for x in plt.gca().get_xticks()]) 
plt.tick_params(axis='both', which='major', labelsize=15)

plt.title("Fit ALMA simulation of point-source source galaxy")
plt.ylabel("Proportions of confidence intervals\ngiving correct predictions", fontsize = 11.5)
plt.xlabel("Percentiles of confidence intervals", fontsize = 11.5)
plt.legend()

plt.show()
```
